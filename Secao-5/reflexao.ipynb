```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# O vetor que receberá a transformação
vetorRE = np.array([[1],[3]])
vetorRE
```

```python
# Matriz transformação
T = np.array([[-1,0],[0,1]])
T
```

```python
# Coordenadas do vetor
x = [0, vetorRE[0][0]]
y = [0, vetorRE[1][0]]
```

```python
plt.plot(x,y, color = 'blue')
```

```python
# Cálculo da transformação
matrizT = np.dot(T,vetorRE)
matrizT
```

```python
xT = [0, matrizT[0][0]]
yT = [0, matrizT[1][0]]
```

```python
plt.plot(x,y, color = 'blue');
plt.plot(xT,yT, color = 'red');
```
