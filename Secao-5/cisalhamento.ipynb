```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Vetor que receberá a transformação de cisalhamento
vetorC = np.array([2, 2])
vetorC
```

```python
# constante de cisalhamento
k = 3 # diferente de zero
```

```python
# Eixo X
# Matriz de transformação
T = np.array([[1,k],[0,1]])
T
```

```python
# Coordenadas do vetor
x = [vetorC[0], vetorC[1]]
y = [0, vetorC[1]]
```

```python
plt.plot(x,y, color = 'blue');
```

```python
# cálculo para a transformação do vetorC
matrizT = np.dot(T, vetorC)
matrizT
```

```python
# coordenadas do vetor transformado

xT = [matrizT[1]+2, matrizT[0]]
yT = [0, matrizT[1]]
```

```python
plt.plot(x,y, color = 'blue');
plt.plot(xT,yT, color = 'red');
plt.show();
```

```python
# Eixo Y
T2 = np.array([[1,0],[k,1]])
T2
```

```python
# Cisalhamento
matrizT2 = np.dot(T2, vetorC)
matrizT2
```

```python
# Coordenadas do vetor

x = [0, vetorC[1]]
y = [vetorC[0], vetorC[1]]

xT2 = [0, matrizT2[0]]
yT2 = [matrizT2[0], matrizT2[1]]
```

```python
plt.plot(x,y, color = 'blue');
plt.plot(xT2,yT2, color = 'red');
```

```python
# Agora vamos tentar outro exemplo
quadrado = np.array([
           [0,0],
           [2,0],
           [2,2],
           [0,2],
           [0,0]
           ])

quadrado_transposto = quadrado.T
quadrado_transposto
```

```python
# 2. Definindo a matriz de cisalhamento (k=3 no eixo X)
k = 2
M_cisalhamento = np.array([
    [1, k],
    [0, 1]
])
M_cisalhamento
```

```python
# 3. Aplicando a transformação
# O .T (transposta) é usado para alinhar as matrizes para multiplicação

paralelogramo = np.dot(M_cisalhamento, quadrado_transposto).T
paralelogramo
```

```python
# 4. Plotagem (Extraindo as linhas: linha 0 é X, linha 1 é Y)
plt.plot(quadrado_transposto[0, :], quadrado_transposto[1, :], 'b--', label='Quadrado')
plt.plot(paralelogramo[:,0], paralelogramo[:,1], 'r-', label='Paralelogramo')

plt.legend(); plt.grid(True); plt.show();
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definindo os pontos (Quadrado e Matriz de Cisalhamento k=3)
q = np.array([[0,0], [2,0], [2,2], [0,2], [0,0]])
p = q @ np.array([[1,0], [3,1]]) # Cisalhamento aplicado
print(p)

# 2. Plotagem extraindo X e Y de cada matriz
plt.plot(q[:,0], q[:,1], 'b--', label='Quadrado') # [x1, x2...], [y1, y2...]
plt.plot(p[:,0], p[:,1], 'r-',  label='Paralelogramo')

plt.legend(); plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Matriz de Transformação (Cisalhamento k=3 no eixo X)
# Vem primeiro no np.dot
M = np.array([[1, 3], 
              [0, 1]])

# 2. Vértices do Quadrado (cada coluna é um ponto [x, y])
# Note o .T para que os pontos fiquem em colunas
quadrado = np.array([[0,0], [2,0], [2,2], [0,2], [0,0]]).T


# 3. Transformação usando np.dot (Matriz * Pontos)
paralelogramo = np.dot(M, quadrado)
print(paralelogramo)

# 4. Plotagem (Extraindo as linhas: linha 0 é X, linha 1 é Y)
plt.plot(quadrado[0, :], quadrado[1, :], 'b--', label='Quadrado')
plt.plot(paralelogramo[0, :], paralelogramo[1, :], 'r-', label='Paralelogramo')

plt.legend(); plt.grid(True); plt.show()
```
