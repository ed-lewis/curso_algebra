```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Vetor que receberá a transformação
vetorH = np.array([[2], [1]])
vetorH
```

```python
# Parâmetro de proporção
k = 3
```

```python
# Matriz de transformação
T = np.array([[k,0],[0,k]])
T
```

```python
# Coordenadas do vetor
x = [vetorH[0][0],0]
y = [0,vetorH[1][0]]
```

```python
plt.plot(x,y, color = 'blue');
plt.show();
```

```python
matrizT = np.dot(T, vetorH)
matrizT
```

```python
xT = [matrizT[0][0],0]
yT = [0,matrizT[1][0]]
```

```python
plt.plot(x,y, color = 'blue');
plt.plot(xT,yT, color = 'red');
plt.show();
```
