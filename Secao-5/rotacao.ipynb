```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Vetor que receberá a transformação
vetorR = np.array([[4],[1]])
vetorR
```

```python
# Ângulo de rotação (30 graus = pi/6 radianos)
alfa = np.pi / 6
```

```python
# Matriz transformação linear (rotação)
T = np.array([[np.cos(alfa), (-np.sin(alfa))],
              [np.sin(alfa), np.cos(alfa)]])
T
```

```python
# Coordenadas do vetorR
x = [0, vetorR[0][0]]
y = [0, vetorR[1][0]]
```

```python
plt.plot(x,y, color = 'blue');
```

```python
# Cálculo para transformação do vetorR
matrizT = np.dot(T, vetorR)
matrizT
```

```python
xT = [0,matrizT[0][0]]
yT = [0,matrizT[1][0]]
```

```python
plt.plot(x,y, color = 'blue');
plt.plot(xT,yT, color = 'red');
plt.show()
```
