```python
import numpy as np
```

```python
#multiplicando matrizes

# Demonstrando a multiplicação de matrizes manualmente

A = np.array([[2,2,3],[3,2,2]])
B = np.array([[1,2,3],[3,2,1],[2,0,1]])
print(A, "\n")
print(B)

resultado = []
for i in range(A.shape[0]):
    linha = []
    print("\n linha", i)
    for j in range(B.shape[0]):
        soma = 0
        print("coluna", j)
        for k in range(A.shape[1]):
            soma += A[i][k] * B[k][j]
        linha.append(soma)
    resultado.append(linha)
```

```python
print(np.array(resultado))
```

```python
R = np.dot(A,B)
R
```
