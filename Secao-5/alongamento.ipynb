```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# vetor que receberá a transformação
vetorA = np.array([4,2])
vetorA
```

```python
# constante = 
# alongamento: k > 1
# compressão: 0 < k < 1

k = 2
```

# Eixo X

```python
# Matriz de transformação
T = np.array([[k, 0], [0, 1]])
T
```

```python
# Coordenadas do vetor
x = [0, vetorA[0]]
y = [vetorA[1], vetorA[1]]
```

```python
plt.plot(x,y, color = 'blue')
```

```python
# Cálculo da transformação
matrizT = np.dot(T, vetorA)
matrizT
```

```python
xT = [0+6, matrizT[0]+6]
yT = [matrizT[1], matrizT[1]]
```

```python
plt.plot(x,y, color = 'blue')
plt.plot(xT,yT, color = 'red')
```

# Eixo Y

```python
# Matriz Transformação
T2 = np.array([[1,0],[0,k]])
T2
```

```python
# Coordenadas do vetor
x = [vetorA[0],vetorA[0]]
y = [0, vetorA[1]]
```

```python
plt.plot(x,y, color = "blue");
```

```python
matrizT2 = np.dot(T2,vetorA)
matrizT2
```

```python
    xT2 = [vetorA[0]+1, vetorA[0]+1]
    yT2 = [0, matrizT2[1]]
```

```python
plt.plot(x, y, color = 'blue', label='Vetor A')
plt.plot(xT2, yT2, color = 'red', label='Vetor A tranformado');
```
