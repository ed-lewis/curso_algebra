Criação de um modelo de regressão linear múltipla \
\
$y = b_0 + b_1x_1 + b_2x_2 + ... + b_nx_n + E$

```python
import numpy as np
import pandas as pd
import scipy as sp
```

```python
# IMPORTAÇÃO DO DATAFRAME
# Substitua 'ID_DO_ARQUIVO' pelo ID real que você copiou do link
file_id = '1w0MhxeKDMbsR_HXVLzxTh8YZZzyWVsw-'
url = f'https://drive.google.com/uc?id={file_id}'           
# O pandas agora consegue acessar o arquivo puramente
nota = pd.read_excel(url)  
resultado = nota
```

```python
resultado.head()
```

```python
A = resultado.drop(columns=['nota'])
A.head()
```

```python
A = A.assign(unidade=1)
A.head()
```

```python
A = A[['unidade', 'inicio_estudo', 'tempo_estudo_dia']]
A.head()
```

```python
b = resultado.drop(columns=['inicio_estudo', 'tempo_estudo_dia'])
b = b.values
b
```

```python
At = np.array(A.T)
At
```

```python
x = np.dot(np.dot(np.linalg.inv(np.dot(At,A)),At),b)
x
```

nota = 6,4871 + 0,0413.inicio_estudo + 0,00644.tempo_estudo_dia

```python
resultado['previsao'] = x[0] + resultado['inicio_estudo']*x[1] + resultado['tempo_estudo_dia']*x[2]
resultado
```

```python
# Erro Médio Absoluto (MAE)
from sklearn.metrics import mean_absolute_error
mean_absolute_error(resultado.nota, resultado.previsao)
```

```python
resultado['erro_abs'] = abs(resultado.nota - resultado.previsao)
resultado
```

```python
import statsmodels.formula.api as smf
import statsmodels.stats.api as sms
```

```python
# Criação do modelo usando a biblioteca Statsmodels
modelo = smf.ols('nota ~ inicio_estudo + tempo_estudo_dia', data = nota).fit()
```
