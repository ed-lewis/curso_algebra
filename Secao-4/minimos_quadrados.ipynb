```python
def produto_interno_vetores(vetor1:np.array,vetor2:np.array) -> float:
    while len(vetor1)<len(vetor2):
        vetor1 = np.append(vetor1,0)
    while len(vetor1)>len(vetor2):
        vetor2 = np.append(vetor2,0)
    return float(np.dot(vetor1, vetor2))
```

```python
v1 = [1,2,3,4]
v2 = [1,2,3]
# Calculamos o tamanho do menor deles usando a função min()
min_len = min(len(v1), len(v2)) 
print(f'O mínimo entre os dois vetores é: {min_len}')
# Aqui, min_len será igual a 3
v_res = np.dot(v1[:min_len], v2[:min_len])
print(f"O produto é: {v_res}")

# Por que isso é eficiente?
# Você não precisou criar um loop while nem adicionar zeros. 
# O "corte" [:min_len] é uma operação extremamente rápida no NumPy, 
# pois ele não copia os dados, apenas cria uma "visão" reduzida do vetor original.
```
